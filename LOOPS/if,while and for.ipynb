#### Ex01:

Write a Python program that asks the user to enter a grade between 0 and 20.
If the grade is less than 10, print "Fail".
If the grade is between 10 and 11.99, print "Pass".
If the grade is between 12 and 15.99, print "Good".
If the grade is 16 or higher, print "Excellent".
If the grade is less than 0 or greater than 20, print "Invalid grade".

```python
n=float(input("enter a number:"))
if n<=0 or n>20 :
    print("invalid")
elif n<10 :
    print("fail")
elif n<=11.99:
    print("pass")
elif n<=15.99:
    print("good")
else :
    print("excellent")
```

#### Ex06:

A salesperson has a daily sales target of 10,000 DA. Use a while loop to ask for the daily sales values:
Accumulate the sales in a variable named total_sales.
Keep asking for sales as long as the total is less than 10,000 DA.
If the user enters a negative value, print 'Invalid sales' and do not add it to the total.
Stop the loop when the total reaches or exceeds the target.
Print the total sales and the number of valid entries required to achieve the target

```python
target=10000
sales=0
i=0
while sales<target:
    n=float(input("enter an amount"))
    i+=1
    sales=n+sales
    if sales<0:
        i-=1
        print("invalid sales")
    else:
        continue       
print("your total is :",sales,"valid enteris are:",i,"target achived")
```

The count of even numbers.
The count of odd numbers.

```python
even=0
seven=0
sodd=0
odd=0
t=0
n=int(input("enter a positive number"))
while n<0:
    n=int(input("enter a positive number"))
for i in range(1,n+1):
    if i%2==0:
        even+=1
        seven=seven+i
    else:
        odd+=1
        sodd=sodd+i
        if i%3==0 and i%5==0:
            t+=1
            break
print(odd,sodd,even,seven,t)
```

#### Ex09:

A program has a fixed secret number set to 37. Ask the user to guess the number. After each attempt:
If the guess is less than the secret number, print 'Too low'.
If it is greater, print 'Too high'.
If it is correct, print 'Correct!'.
Add a trial system: The user has 6 attempts. After each wrong guess, decrease the remaining number of attempts. If they find the number, stop the game. If attempts run out, print the secret number.

```python
secret=37
i=6
while i>0:
    i-=1
    n=int(input("guess the number"))
    if secret>n :
        print("too low")
    elif secret<n:
        print("too high")
    else :
        print("its correct")
        break
print("stop the game its:",secret)
```
